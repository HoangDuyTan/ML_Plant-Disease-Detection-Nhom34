```python
# TEST
import os
import matplotlib.pyplot as plt
import matplotlib.image as mpimg

# 1. Tự động quét toàn bộ thư mục Kaggle để tìm đúng thư mục 'train' và 'valid'
train_dir = None
valid_dir = None

for root, dirs, files in os.walk('/kaggle/input'):
    if root.endswith('train'):
        train_dir = root
    elif root.endswith('valid'):
        valid_dir = root

print(f"✅ Đã tìm thấy đường dẫn Train: {train_dir}")

# 2. Xử lý dữ liệu sau khi tìm thấy đường dẫn
if train_dir is None:
    print("❌ Vẫn chưa tìm thấy! Hãy kiểm tra lại xem đã Add Dataset ở cột phải chưa nhé.")
else:
    # Đếm số lượng thư mục (chính là số loại bệnh)
    classes = os.listdir(train_dir)
    num_classes = len(classes)
    print(f"🔥 Báo cáo Nhóm 34: Đã tìm thấy {num_classes} loại bệnh khác nhau trong tập Train!")

    print("\nDanh sách 5 loại bệnh đầu tiên:")
    for i in range(5):
        print(f"- {classes[i]}")

    # 3. Rút đại 1 bức ảnh trong thư mục đầu tiên ra xem mặt mũi nó thế nào
    sample_class = classes[0]
    sample_dir = os.path.join(train_dir, sample_class)
    
    # Lấy tên file ảnh đầu tiên
    sample_image_name = os.listdir(sample_dir)[0]
    sample_image_path = os.path.join(sample_dir, sample_image_name)

    # Vẽ ảnh lên màn hình
    img = mpimg.imread(sample_image_path)
    plt.figure(figsize=(4, 4))
    plt.imshow(img)
    plt.title(f"Minh họa: {sample_class}")
    plt.axis('off')
    plt.show()
```
