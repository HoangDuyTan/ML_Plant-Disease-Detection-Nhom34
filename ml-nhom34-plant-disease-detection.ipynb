```python
train_dir = glob.glob('/kaggle/input/**/train', recursive=True)[0]
valid_dir = glob.glob('/kaggle/input/**/valid', recursive=True)[0]
```

**Phần của Tân:**

```python
# Data Preprocessing
IMG_SIZE = (224, 224)
BATCH_SIZE = 32

train_dts = utils.image_dataset_from_directory(
    train_dir,
    shuffle=True,
    batch_size=BATCH_SIZE,
    image_size=IMG_SIZE,
    label_mode='categorical'
)

valid_dts = utils.image_dataset_from_directory(
    valid_dir,
    shuffle=False,
    batch_size=BATCH_SIZE,
    image_size=IMG_SIZE,
    label_mode='categorical'
)

class_names = train_dts.class_names
print(f"Tiền xử lý dữ liệu thành công, đã nạp {len(class_names)} loại bệnh")
```
